### Cleaning validation function

```python
def validate_clean_data(data):
    assert data.shape[0] == 7043, "Unexpected number of rows"

    assert data["customerID"].is_unique, \
        "customerID contains duplicates"

    assert data["TotalCharges"].isna().sum() == 0, \
        "TotalCharges still contains missing values"

    assert pd.api.types.is_numeric_dtype(
        data["TotalCharges"]
    ), "TotalCharges is not numeric"

    assert data["Churn"].isin(["Yes", "No"]).all(), \
        "Unexpected value found in Churn"

    assert (data["tenure"] >= 0).all(), \
        "Negative tenure found"

    assert (data["MonthlyCharges"] >= 0).all(), \
        "Negative MonthlyCharges found"

    assert (data["TotalCharges"] >= 0).all(), \
        "Negative TotalCharges found"

    print(" All data quality checks passed.")
```

```python
validate_clean_data(df)
```

```python
# clean dataset save
OUTPUT_PATH = Path(
    "../data/processed/telco_churn_clean.csv"
)

df.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"Clean dataset saved to: {OUTPUT_PATH}")
```
